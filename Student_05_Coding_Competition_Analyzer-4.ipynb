## 1. Import libraries
NumPy is used for numerical calculations. Pandas is used to read, organise, filter and analyse table data.


```python
import numpy as np
import pandas as pd
```

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
# Load the CSV file (keep it in the same folder as this notebook)
df = pd.read_csv("Student_05_dataset.csv")

print("Dataset loaded successfully!")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
# Display the number of rows and columns
print("Dataset shape:", df.shape)

# Display column names
print("\nColumn names:")
print(df.columns.tolist())

# Inspect data types and non-null values
print("\nDataset information:")
df.info()

# Count missing values in each column
print("\nMissing values:")
print(df.isnull().sum())

# Count duplicate rows
print("\nNumber of duplicate rows:", df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# Select the numeric columns relevant to the competition
numeric_columns = ["Problems_Attempted", "Problems_Solved", "Score"]
numeric_df = df[numeric_columns]

# Convert the selected columns into a NumPy array
numeric_array = numeric_df.to_numpy()


# Calculate average, maximum and minimum for each selected column
print("\nAverage values:")
print(np.mean(numeric_array, axis=0))

print("\nMaximum values:")
print(np.max(numeric_array, axis=0))

print("\nMinimum values:")
print(np.min(numeric_array, axis=0))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# Make a copy for analysis so the original DataFrame remains unchanged
analysis_df = df.copy()

# Calculated column 1: Solve rate in percentage
analysis_df["Solve_Rate"] = np.where(
    analysis_df["Problems_Attempted"] > 0,
    (analysis_df["Problems_Solved"] / analysis_df["Problems_Attempted"]) * 100,
    0
).round(2)

# Calculated column 2: Problems not solved
analysis_df["Problems_Not_Solved"] = (
    analysis_df["Problems_Attempted"] - analysis_df["Problems_Solved"]
)

print("Data with calculated columns:")
display(analysis_df.head())

# Filter students whose solve rate is 50% or more
print("\nStudents with a solve rate of 50% or more:")
high_solve_rate = analysis_df[analysis_df["Solve_Rate"] >= 50]
display(high_solve_rate)

# Sort students by score in descending order
print("\nTop 5 students by score:")
sorted_students = analysis_df.sort_values(by="Score", ascending=False)
display(sorted_students.head(5))

# Display the 5 lowest scores
print("\nBottom 5 students by score:")
display(sorted_students.tail(5))
```

## 6. Your individual project challenge
**Challenge:** Calculate solve rate; create leaderboard; identify top five scorers.

Write your solution below and explain what the result means.


```python
# Create the competition leaderboard
leaderboard = df.copy()

# Calculate solve rate for each student
leaderboard["Solve_Rate"] = np.where(
    leaderboard["Problems_Attempted"] > 0,
    (leaderboard["Problems_Solved"] / leaderboard["Problems_Attempted"]) * 100,
    0
).round(2)

# Sort by score from highest to lowest
leaderboard = leaderboard.sort_values(
    by="Score", ascending=False
).reset_index(drop=True)

# Assign ranks starting from 1
leaderboard.insert(0, "Rank", range(1, len(leaderboard) + 1))

print("Complete Coding Competition Leaderboard:")
display(leaderboard)

# Display the top five scorers
print("\nTop 5 scorers:")
display(leaderboard.head(5))

# Display overall competition statistics
print("\nCompetition Summary")
print("Total students:", len(leaderboard))
print("Total problems attempted:", leaderboard["Problems_Attempted"].sum())
print("Total problems solved:", leaderboard["Problems_Solved"].sum())
print("Average score:", round(leaderboard["Score"].mean(), 2))
print("Average solve rate:", round(leaderboard["Solve_Rate"].mean(), 2), "%")
```
